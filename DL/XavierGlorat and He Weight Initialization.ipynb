```python
from sklearn.datasets import make_moons
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
X,y =make_moons(100, noise=0.25,random_state=2)
```

```python
plt.scatter(X[:,0],X[:,-1],c=y)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(10,activation='relu',input_dim=2,kernel_initializer='he_normal'))
model.add(Dense(10,activation='relu',kernel_initializer='he_normal'))
model.add(Dense(10,activation='relu',kernel_initializer='he_normal'))
model.add(Dense(10,activation='relu',kernel_initializer='he_normal'))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model.fit(X,y,epochs=100,validation_split=0.2)
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype('int'),clf=model,legend=2)
```

```python
model2 = Sequential()
model2.add(Dense(10,activation='tanh',input_dim=2,kernel_initializer='glorot_normal'))
model2.add(Dense(10,activation='tanh',kernel_initializer='glorot_normal'))
model2.add(Dense(10,activation='tanh',kernel_initializer='glorot_normal'))
model2.add(Dense(10,activation='tanh',kernel_initializer='glorot_normal'))
model2.add(Dense(1,activation='sigmoid'))
model2.summary()
```

```python
model2.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model2.fit(X,y,epochs=100,validation_split=0.2)
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype('int'),clf=model2,legend=2)
```
