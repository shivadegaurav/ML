```python
from keras.applications.vgg16 import VGG16
model = VGG16()
```

```python
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model)
```

```python
for i in range(len(model.layers)):
    #check for convolutional layer 
    if 'conv' not in model.layers[i].name:
        continue
    #get filter weights
    filters, biases = model.layers[i].get_weights()
    print("layer number",i,model.layers[i].name,filters.shape)
```

```python
filters,bias = model.layers[1].get_weights()
```

```python
f_min,f_max = filters.min(),filters.max()
filters = (filters - f_min) / (f_max - f_min)
```

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt

n_filters = 6
ix = 1
fig = plt.figure(figsize=(15,10))
for i in range(n_filters):
    # get the filters
    f = filters[:,:,:,i]
    for j in range(3):
        plt.subplot(n_filters, 3, ix)
        plt.imshow(f[:,:,j], cmap='gray')
        plt.axis('off')   # optional, hides tick marks
        ix += 1
# plot the filters
plt.show()
```

```python
from tensorflow.keras.models import Model
```

```python
model = Model(inputs=model.inputs,outputs=model.layers[1].output)
```

```python
from tensorflow.keras.applications.vgg16 import preprocess_input
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from numpy import expand_dims
image = load_img('tiger.jpg',target_size=(224,224))
image = img_to_array(image)
image = expand_dims(image,axis=0)
image = preprocess_input(image)
```

```python
import matplotlib.pyplot as plt

features = model.predict(image)
fig = plt.figure(figsize=(20,15))
for i in range(1, features.shape[3] + 1):
    plt.subplot(8, 8, i)
    plt.imshow(features[0, :, :, i-1], cmap='gray')
    plt.axis('off')
plt.show()
```

```python
model2 = VGG16()
```

```python
layer_index = [2,5,9,13,17]
outputs = [model2.layers[i].output for i in layer_index]
model3 = Model(inputs = model2.inputs,outputs = outputs)
```

```python
import matplotlib.pyplot as plt

feature_map = model3.predict(image)

for i, fmap in zip(layer_index, feature_map):
    fig = plt.figure(figsize=(20,15))
    fig.suptitle("Layer_{}".format(i), fontsize=20)  # use suptitle, not subtitle
    for j in range(1, fmap.shape[3] + 1):
        plt.subplot(8, 8, j)
        plt.imshow(fmap[0, :, :, j-1], cmap='gray')
        plt.axis('off')
    plt.show()
```
